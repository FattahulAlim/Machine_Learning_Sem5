```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import tensorflow as tf
```

```python
# Load dataset
iris = load_iris()
X = iris.data
y = iris.target.reshape(-1, 1)

# One-hot encoding
encoder = OneHotEncoder(sparse_output=False)
y = encoder.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Bangun model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
model.fit(X_train, y_train, epochs=50, batch_size=8)

# Evaluasi
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi: {acc}")
```

# Tugas 2

```python
def latih_model(n1, n2, epochs=50, batch_size=8):
    tf.keras.utils.set_random_seed(42)  # agar bobot awal sama
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(n1, activation='relu', input_shape=(4,)),
        tf.keras.layers.Dense(n2, activation='relu'),
        tf.keras.layers.Dense(3, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, verbose=0)
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    return loss, acc

# Sebelum: konfigurasi awal (10 dan 8 neuron)
loss_awal, acc_awal = latih_model(10, 8)

# Sesudah: jumlah neuron diubah (20 dan 16 neuron)
loss_baru, acc_baru = latih_model(20, 16)

print(f"=== Sebelum (10-8) ===")
print(f"Loss:{loss_awal:.4f}")
print(f"Akurasi: {acc_awal:.4f}")
print()
print("=== Sesudah (20-16) ===")
print(f"Loss:{loss_baru:.4f}")
print(f"Akurasi: {acc_baru:.4f}")
```

## Perbandingan Konfigurasi Hidden Layer pada Dataset Iris

### 1. Hasil Evaluasi

| Konfigurasi                  | Jumlah Parameter | Loss (Test) | Akurasi (Test) |
|------------------------------|------------------|-------------|----------------|
| Sebelum (Hidden 10 - 8)      | 165              | 0.1809      | 0.9667         |
| Sesudah (Hidden 20 - 16)     | 487              | 0.1006      | 0.9667         |

### 2. Analisis

- **Akurasi:** Kedua konfigurasi menghasilkan akurasi yang sama, yaitu 0.9667. Jadi menambah jumlah neuron tidak meningkatkan akurasi pada percobaan ini.
- **Loss:** Loss konfigurasi 20-16 (0.1006) lebih kecil daripada konfigurasi 10-8 (0.1809). Artinya model dengan neuron lebih banyak memberikan probabilitas yang lebih yakin pada kelas yang benar, meskipun jumlah prediksi benarnya sama.
- **Kesimpulan:** Untuk dataset Iris yang kecil dan sederhana, konfigurasi 10-8 sudah cukup untuk mencapai akurasi yang sama. Menambah neuron memberi loss yang lebih kecil, tetapi tidak menambah akurasi. Karena data test hanya 30 sampel, perbedaan kecil pada akurasi belum tentu mencerminkan perbedaan kualitas model yang sebenarnya.

# Tugas 3

```python
def latih_aktivasi(aktivasi, epochs=50, batch_size=8):
    tf.keras.utils.set_random_seed(42)  # agar bobot awal sama
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(10, activation=aktivasi, input_shape=(4,)),
        tf.keras.layers.Dense(8, activation=aktivasi),
        tf.keras.layers.Dense(3, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, verbose=0)
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    return loss, acc

loss_sig, acc_sig = latih_aktivasi('sigmoid')
loss_relu, acc_relu = latih_aktivasi('relu')

print("=== Sigmoid ===")
print(f"Loss: {loss_sig:.4f}")
print(f"Akurasi: {acc_sig:.4f}")
print()
print("=== ReLU ===")
print(f"Loss: {loss_relu:.4f}")
print(f"Akurasi: {acc_relu:.4f}")
```

## Perbandingan Fungsi Aktivasi Sigmoid vs ReLU pada Dataset Iris

### 1. Hasil Evaluasi

| Fungsi Aktivasi | Loss (Test) | Akurasi (Test) |
|-----------------|-------------|----------------|
| Sigmoid         | 0.6111      | 0.7667         |
| ReLU            | 0.1809      | 0.9667         |

### 2. Analisis

- **Akurasi:** ReLU mencapai akurasi 0.9667, sedangkan Sigmoid hanya 0.7667.
- **Loss:** Loss ReLU (0.1809) jauh lebih kecil dibandingkan Sigmoid (0.6111), sekitar 70% lebih rendah. Artinya prediksi ReLU lebih mendekati label sebenarnya.
- **Kemungkinan penyebab:** Turunan Sigmoid nilainya kecil (maksimal 0.25) dan makin mengecil saat neuron jenuh, sehingga gradien yang dirambatkan ke layer sebelumnya melemah (*vanishing gradient*) dan proses belajar jadi lambat. Pada ReLU, gradien bernilai 1 untuk input positif sehingga belajar lebih cepat. Dengan jumlah epoch yang sama (50), Sigmoid kemungkinan belum sempat konvergen.
- **Kesimpulan:** Pada dataset Iris dengan konfigurasi dan jumlah epoch yang sama, ReLU memberikan hasil yang jauh lebih baik dibandingkan Sigmoid, baik dari sisi loss maupun akurasi.
