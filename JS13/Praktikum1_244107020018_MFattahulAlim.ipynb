```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

```python
# Mengubah hidden layer menjadi 3

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 3 # Mengubah Hidden layer menjadi 3
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

- **Loss akhir:** Di epoch 9000, loss hidden 3 neuron (0.0024) lebih kecil daripada hidden 2 neuron (0.0031).
- **Hasil prediksi:** Kedua konfigurasi sama-sama berhasil mempelajari pola XOR. Output mendekati 0 untuk input [0,0] dan [1,1], serta mendekati 1 untuk input [0,1] dan [1,0].
- **Kesimpulan:** Menambah neuron pada hidden layer membuat model punya kapasitas lebih besar, sehingga training lebih cepat dan loss akhir lebih kecil. Namun untuk kasus sederhana seperti XOR, 2 neuron sebenarnya sudah cukup. Perbedaannya lebih terasa di kecepatan belajar daripada akurasi akhir.

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(a):      # input: output sigmoid (a)
    return a * (1 - a)

def relu(x):
    return np.maximum(0, x)

def relu_derivative(z):         # input: z (sebelum aktivasi)
    return (z > 0).astype(float)

def train(hidden_size, activation="sigmoid", lr=0.1, epochs=10000):
    W1 = np.random.randn(2, hidden_size)
    b1 = np.zeros((1, hidden_size))
    W2 = np.random.randn(hidden_size, 1)
    b2 = np.zeros((1, 1))

    for epoch in range(epochs):
        # Forward
        z1 = np.dot(X, W1) + b1
        a1 = sigmoid(z1) if activation == "sigmoid" else relu(z1)
        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2)

        error = y - a2

        # Backprop
        d_a2 = error * sigmoid_derivative(a2)
        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        if activation == "sigmoid":
            d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
        else:
            d_a1 = np.dot(d_a2, W2.T) * relu_derivative(z1)
        d_W1 = np.dot(X.T, d_a1)
        d_b1 = np.sum(d_a1, axis=0, keepdims=True)

        # Update
        W1 += lr * d_W1
        b1 += lr * d_b1
        W2 += lr * d_W2
        b2 += lr * d_b2

        if epoch % 1000 == 0:
            print(f"Epoch {epoch}, Loss: {np.mean(np.square(error))}")

    return a2

print("=== Hidden 3, Sigmoid ===")
print(train(3, "sigmoid"))

print("=== Hidden 3, ReLU ===")
print(train(3, "relu"))
```

### 3. Analisis

- **Konvergensi:** Pada Sigmoid, loss turun bertahap dan stabil sampai 0.0026 di epoch 9000. Pada ReLU, loss sempat turun cepat di awal (0.3724 ke 0.1672 pada epoch 1000), tetapi setelah itu stagnan di sekitar 0.1667 dan tidak turun lagi sampai akhir training.
- **Loss akhir:** Loss akhir Sigmoid (0.0026) jauh lebih kecil dibandingkan ReLU (0.1667).
- **Hasil prediksi:** Sigmoid berhasil mempelajari pola XOR, dengan output mendekati 0 untuk [0,0] dan [1,1] serta mendekati 1 untuk [0,1] dan [1,0]. ReLU gagal, karena tiga input pertama menghasilkan output yang sama (0.6665) sehingga model tidak bisa membedakan input.
- **Penyebab kegagalan ReLU:** Kemungkinan besar terjadi masalah *dying ReLU*. Jika input suatu neuron bernilai negatif, output ReLU menjadi 0 dan gradiennya juga 0, sehingga bobot neuron tersebut tidak diperbarui lagi. Karena bobot awal dibuat acak dan jumlah neuron hanya 3, beberapa neuron mati sejak awal dan model kehilangan kapasitas untuk memisahkan pola XOR.
- **Kesimpulan:** Pada percobaan ini, Sigmoid lebih stabil dan lebih andal untuk kasus XOR dengan hidden layer kecil. ReLU sensitif terhadap inisialisasi bobot awal sehingga bisa gagal konvergen.
