```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
import time
```

```python
# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model JST
model = Sequential([
    Flatten(input_shape=(28, 28)),       # Ubah gambar 28x28 menjadi vektor 784
    Dense(128, activation='relu'),       # Hidden layer 1
    Dense(64, activation='relu'),        # Hidden layer 2
    Dense(10, activation='softmax')      # Output layer (10 kelas)
])

# 3. Kompilasi model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 4. Latih model
model.fit(X_train, y_train, epochs=10, batch_size=32)

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
```

```python
def eksperimen(hidden_layers, aktivasi='relu', epochs=10):
    tf.keras.utils.set_random_seed(42)  # bobot awal sama

    model = Sequential()
    model.add(Flatten(input_shape=(28, 28)))
    for n in hidden_layers:
        model.add(Dense(n, activation=aktivasi))
    model.add(Dense(10, activation='softmax'))

    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    mulai = time.time()
    model.fit(X_train, y_train, epochs=epochs, batch_size=32, verbose=0)
    waktu = time.time() - mulai

    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    return acc, loss, waktu

konfigurasi = [
    ("Awal (128-64), ReLU",              [128, 64],      'relu'),
    ("Neuron lebih banyak (256-128)",    [256, 128],     'relu'),
    ("3 hidden layer (256-128-64)",      [256, 128, 64], 'relu'),
    ("Awal (128-64), Sigmoid",           [128, 64],      'sigmoid'),
]

print(f"{'Konfigurasi':<34}{'Akurasi':<10}{'Loss':<10}{'Waktu (detik)'}")
print("-" * 68)
for nama, layers, akt in konfigurasi:
    acc, loss, waktu = eksperimen(layers, akt)
    print(f"{nama:<34}{acc:<10.4f}{loss:<10.4f}{waktu:.1f}")
```
