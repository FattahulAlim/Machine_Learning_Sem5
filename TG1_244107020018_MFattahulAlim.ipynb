1. Contoh pelanggaran etika dan hukum tentang penggunaan kecerdasan buatan :     

**Contoh Kasus: Bias Gender pada Algoritma Rekrutmen Amazon**

Salah satu contoh nyata pelanggaran etika sekaligus pelanggaran hukum dalam penggunaan kecerdasan buatan adalah kegagalan sistem rekrutmen eksperimental berbasis AI yang dikembangkan oleh Amazon. Sistem ini dirancang untuk menyortir CV pelamar secara otomatis dengan memberikan peringkat satu hingga lima bintang menggunakan model yang dilatih dari 50.000 istilah yang diekstrasi dari data historis CV selama 10 tahun (Kodiyan, 2019; Haughey, Mann, & Oca, 2025).


Permasalahan muncul ketika data pelatihan didominasi oleh CV pelamar laki-laki sebagai representasi dari ketimpangan gender struktural yang sudah lama ada. Akibatnya algoritma menurunkan peringkat resume yang memuat kata kata terkait perempuan seperti "women's Chess club captain", serta mendiskriminasi lulusan dari perguruan tinggi khusus perempuan (Kodiyan, 2019; Haughey, Mann, & Oca, 2025). Sehingga, amazon akhirnya menghentikan penggunaan sistem tersebut

Kasus di atas melanggar beberapa prinsip etika dan hukum dalam penggunaan AI di antaranya:


*   Fairness / Keadilan (Pelanggaran Etika): Algoritma gagal memberikan kesempatan yang setara bagi kelompok yang terpinggirkan. Walaupun data gender pelamar tidak dimasukkan secara langsung, algoritma tetap bisa menebak gender kandidat lewat variabel lain yang berhubungan, seperti jeda masa kerja, gaya bahasa, dan kegiatan ekstrakurikuler (Yu, 2025).
*   Akuntabilitas dan Transparansi (Pelanggaran Etika): Model machine learning bekerja seperti "kotak hitam" (black box) yang sulit ditelusuri cara pengambilan keputusannya. Akibatnya, sulit mencari tahu siapa yang harus bertanggung jawab saat terjadi diskriminasi (Silva, 2024; Yu, 2025).
*   Diskriminasi Ketenagakerjaan: Praktik ini berpotensi melanggar prinsip kesetaraan kesempatan kerja (equal employment opportunity) yang diatur dalam berbagai peraturan ketenagakerjaan di banyak negara, karena secara sistematis merugikan kandidat berdasarkan gender.
*   Paradoks Regulasi Privasi (GDPR Paradox): Regulasi seperti GDPR di Eropa melarang pengumpulan data sensitif seperti gender demi melindungi privasi individu. Namun, aturan ini justru menimbulkan masalah baru: tanpa data gender, developer jadi kesulitan memeriksa dan membuktikan adanya bias pada algoritma. Alhasil, hukum yang seharusnya melindungi malah menghambat upaya mendeteksi diskriminasi (Yu, 2025).

2. Dampak energi dan lingkungan terhdap pemanfaatan kecerdasan buatan :     

**Contoh dampak : Konsumsi Air Data Center AI di Indonesia**

Salah satu dampak lingkungan yang terjadi akibat pemanfaatan kecerdasan buatan adalah tingginya konsumsi air untuk mendinginkan server data center. Di indonesia ekspansi infrastruktur dara center AI berlangsng dengan sangat cepat, dengan nilai pasara mencapai 3.7 miliar us dollar pada tahub 2024 dan terus bertumbuh hingga 5.8 milliar us dollar pada tahun 2030 (Bakhriyah, 2026).

Masalahnya menjadi lebih kompleks akibat iklim tropis Indonesia dengan suhu rata rata 27 sampai 35 derajat dan kelembapan tinggi (70-90%), sehingga menurunkan efisiensi sistem pendingin evaporatif dibandingkan negara beriklim sedang. Akibatnya cooling tower harus mensirkulasikan volume air yang jauh lebih besar untuk mencapai efek pendinginan yang setara (Bakhriyah, 2026). Dampak ini terlihat nyata di wilayah-wilayah konsentrasi data center seperti koridor Jakarta-Bekasi-Cikarang, yang menampung lebih dari 55% kapasitas data center nasional namun mengalami penurunan muka tanah 10-15 cm per tahun akibat eksploitasi air tanah berlebihan. Situasi serupa terjadi di Batam, di mana kapasitas suplai air bersih (PAM) sudah terserap 90% untuk kebutuhan industri dan rumah tangga (Bakhriyah, 2026).

Cara mengatasi:
* Migrasii ke teknologi pendingin hemat air: Teknologi seperti liquid cooling dan immersion cooling contohnya NeutraDC di batam mengklaim mampu menghemat konsumsi air hingga 70% melalu penerapan liquid cooling (Bakhriyah, 2026).
* Pemanfaat sumber daya alam lokal: Indonesia memiliki curah hujan tinggi yang dapat dimanfaatkan melalui rainwater harvesting sebagai sumber air pendingin non-potable, serta cadangan panas bumi terbesar di dunia (40% cadangan global) yang berpotensi digunakan untuk geothermal cooling dengan konsumsi air mendekati nol (Bakhriyah, 2026).
* Penguatan kerangka regulasi: Diperlukan kewajiban pelaporan Water Usage Effectiveness (WUE) yang terstandarisasi bagi operator data center, disertai penetapan target WUE maksimum secara bertahap, serta insentif fiskal bagi fasilitas yang mengadopsi teknologi pendingin hemat air (Bakhriyah, 2026).
