```python
#inspeksi visual

#inspeksi distribusi data
num_cols = ['Age', 'Diabetes_012']
for col in num_cols:
    plt.figure()
    plt.hist(df[col], bins=30)
    plt.title(f'Distribusi {col}')
    plt.xlabel(col); plt.ylabel('Frekuensi')
    plt.show()
```

```python
#Inspeksi Outlier
plt.figure()
plt.boxplot(df['Age'].dropna(), vert=True)
plt.title('Boxplot Age')
plt.ylabel('Age')
plt.show()
```

```python
#Inspeksi distribusi variabel
cat_cols = ['Age','HighChol','Smoker','Diabetes_012']

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    df[col].value_counts(dropna=False).plot(kind='bar', ax=axes[i])
    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Jumlah')

plt.tight_layout()
plt.show()
```

```python
# Heatmap
import seaborn as sns # Import seaborn for enhanced visualization

num_only = df.select_dtypes(include=[np.number])
corr = num_only.corr(numeric_only=True)

plt.figure(figsize=(20, 20)) # Increased figure size for better readability
sns.heatmap(corr, cmap='coolwarm', vmin=-1, vmax=1, annot=True, fmt=".2f") # Use seaborn.heatmap with annotation
plt.title('Heatmap Korelasi (numerik)')
plt.tight_layout()
plt.show()
```

```python
#Inspeksi korelasi dengan hasil data tabular
corr['Age'].sort_values(ascending=False)
```

```python
# Scatter plot
plt.figure()
surv_map = {0:'No', 1:'Yes'}
plt.scatter(df['BMI'], df['Diabetes_012'], alpha=0.7)
plt.xlabel('BMI'); plt.ylabel('Diabetes_012')
plt.title('Scatter BMI vs Diabetes_012')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Membaca gambar (BGR default di OpenCV)
img = cv2.imread("/content/drive/MyDrive/ml-sem5/lena.jpg")

# Konversi BGR → RGB (agar warnanya tampil benar di matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tampilkan
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis("off")
plt.show()
```

```python
# Resize gambar
img_resized = cv2.resize(img_rgb, (128, 128))

plt.imshow(img_resized)
plt.title("Gambar setelah Resize (128x128)")
plt.axis("off")
plt.show()
```

```python
# Konversi ke grayscale
img_gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

plt.imshow(img_gray, cmap="gray")
plt.title("Gambar Grayscale")
plt.axis("off")
plt.show()
```
